```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math as mt

df_i = pd.read_csv('data_set_ALL_AML_independent.csv')
df_t = pd.read_csv('data_set_ALL_AML_train.csv')

display(df_i.head())
display(df_t.head())
```

```python
def perfilado_general(df):
    n, m = df.shape
    print(f'Número de registros: {n}')
    print(f'Número de columnas: {m}')
    print('-'*100)
    print(f'Columnas disponibles: {df.columns}\n')
    display(df.info())
    print('-'*100)
    print(f'Valores nulos: {df.isna().sum().sum()}')
    print(f'Valores duplicados: {df.duplicated().sum()}')
    print('Distribución de datos:')
    for column in df.select_dtypes(include = ('object', 'str')).columns:
        print(f'Valores únicos para la columna {column}: {len(df[column].unique())}')
    col_num = len(df.select_dtypes(include = ('float64', 'int64')).columns)
    plt.figure(figsize = (10, col_num*0.8))
    for i, column in enumerate(df.select_dtypes(include = ('float64', 'int64')).columns):
        plt.subplot(mt.ceil(col_num / 4), 4, i+1)
        plt.boxplot(df[column])
        plt.title(column)
    plt.tight_layout()
    plt.show()
    display(df.describe())
```

```python
perfilado_general(df_i)
```

```python
perfilado_general(df_t)
```
